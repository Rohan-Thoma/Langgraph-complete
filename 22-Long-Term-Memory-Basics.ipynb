## Long Term Memory Basics in Langgraph


First import the required module

```python
from langgraph.store.memory import InMemoryStore
```

Now lets create the memory store first

```python
store = InMemoryStore()
```

Now, lets define a name-space

Here NameSpace is nothing but it acts like a folder inside the memory store that we are creating.

Here , the number of names we give while creating the namespace = number of folders.

namespace = ("user", "u1") 

Here, this means at a time, we are creating 2 subfolder like user and inside it u1.

Now to create another folder inside user, we simply need ("user","u2"), now this will automatically check if the user folder exists and it will add u2 in that.

This is applicable for as many folder names we give, it will represent that many levels of folder organization structure.

```python
namespace = ("user", "u1")
```

Now, lets create memories

1. Adding a memory

We add memories with the help of a put() function , which requires 2 arguments which are : "namespace", "key" and a "value".

namespace for telling it which folder to store the memories in.
As the data is stored in the form of a dictionary, we will giving it a key and value here.

Now the value can also be anything like it even can be another dictionary.

```python
store.put(namespace, "1", {"data": "User likes pizza"})
store.put(namespace, "2", {"data":"User prefers dark mode"})
```

Now, lets add another folder for user2

```python
#Lets add another folder for user2
namespace2= ("user", "u2")

store.put(namespace2, "1", {"data": "User likes pasta"})
store.put(namespace2, "2", {"data": "User prefers grid style navigation"})
```

2. Retrieving the Memories

This is done with the help of get() method, this requires 2 arguments only which is the "namespace" and "key", because we are simply fetching what is already stored etc.

```python
store.get(namespace, "1")
```

2. a. Retrieving all Memories

This is done with the help of search() method, this simply takes one argument  which is "namespace", to get all the items stored in that folder.

```python
items = store.search(namespace)

for item in items:
    print(item.value)
```

2. b Semantic Search for memories.

Now, in the practical application, we will neither fetch the memeories with some particular keyword or neither we will fetch all the memories.

We will need to fetch the memories related to the particular context present before us.

* Now to do this, while creating the store itself, we will give it the ability to store the memories along with their embeddings by passing an embedding model to it via the parameter called "index".

```python
from langchain_openai import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv()

embedding_model = OpenAIEmbeddings(model = 'text-embedding-3-small')

#Now define the store, passing another parameter called "index"
store = InMemoryStore(index = {'embed': embedding_model, 'dims': 1536})
```

Now , lets add memories to test it

```python
namespace = ("users", "u1")

store.put(namespace, "1", {"data": "User prefers concise answers over long explanations"})
store.put(namespace, "2", {"data": "User likes examples in Python"})
store.put(namespace, "3", {"data": "User usually works late at night"})
store.put(namespace, "4", {"data": "User prefers dark mode in applications"})
store.put(namespace, "5", {"data": "User is learning machine learning"})
store.put(namespace, "6", {"data": "User dislikes overly theoretical explanations"})
store.put(namespace, "7", {"data": "User prefers step-by-step reasoning"})
store.put(namespace, "8", {"data": "User is based in India"})
store.put(namespace, "9", {"data": "User likes real-world analogies"})
store.put(namespace, "10", {"data": "User prefers bullet points over paragraphs"})
```

Now, lets search with the help of semantic search.

normal search() function, when we pass the question with "query" parameter.

```python
items = store.search(namespace, query= "what is the user currently learning", limit = 1)

for item in items:
    print(item.value)
```
